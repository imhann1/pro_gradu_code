# Manual Assessment
This notebook includes collection of subsets for the manual assessments of professional domain advice-seeking prompts and model responses.

```python
from huggingface_hub import login
from datasets import load_dataset, concatenate_datasets
import numpy as np
import pandas as pd
```

```python
login()
```

```python
# get datasets from hf
ds_llama = load_dataset("imhann/qwen_evals_llama", split='train')
ds_gpt = load_dataset("imhann/qwen_evals_gpt", split='train')
```

```python
# turn datasets into pandas dataframes
df_llama = ds_llama.to_pandas()
df_gpt = ds_gpt.to_pandas()

gpt_labels = df_gpt[['index', 'evaluation_label', 'prompt_type', 'model_response', 'input']].rename(
    columns={'evaluation_label': 'GPT_category', 'model_response': 'GPT_response'})

llama_labels = df_llama[['index', 'evaluation_label', 'model_response']].rename(
    columns={'evaluation_label': 'Llama_category', 'model_response': 'Llama_response'})

# merge gpt and llama with prompt id
df = pd.merge(
    gpt_labels,
    llama_labels,
    on='index',
    how='inner'
)
```

```python
df
```

# 1. Medical prompts and their responses

28.2% of medical prompts where Llama answered with category 0, GPT answered with category 3. This share is significant, considering 0 and 3 they are on opposite ends of response behavior.

```python
# prompts and responses where Llama label is 0 and GPT label is 3 in medical domain
df_subsection = df[(df['Llama_category'] == 0) & (df['GPT_category'] == 3) & (df['prompt_type'] == 'medical')]
for index, row in df_subsection.iterrows():
    print(f"Prompt: {row['input']}")
    print(f"GPT Response: {row['GPT_response']}")
    print(f"Llama Response: {row['Llama_response']}")
    print("\n")
```
