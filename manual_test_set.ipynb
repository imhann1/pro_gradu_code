This notebook contains the code for collecting Llama's and GPT's responses for the manually evaluated test set. The resulting test set then includes 240 manually annotated examples, comprising 120 GPT responses and 120 Llama responses to the same 120 advice-seeking prompts. The test set
is balanced across the four professional domains included in the prompt dataset, meaning 30 examples were randomly sampled from each professional domain.

# Test Set Creation

```python
from datasets import load_dataset, concatenate_datasets
from huggingface_hub import login
```

```python
login()
```

```python
# load the llama datasets
ds1 = load_dataset('imhann/medical_prompts_responses_llama', split='train')
ds2 = load_dataset('imhann/legal_prompts_responses_llama', split='train')
ds3 = load_dataset('imhann/finance_prompts_responses_llama', split='train')
ds4 = load_dataset('imhann/therapy_prompts_responses_llama', split='train')
# add column to identify 'prompt_type'
ds1 = ds1.add_column('prompt_type', ['medical'] * len(ds1))
ds2 = ds2.add_column('prompt_type', ['legal'] * len(ds2))
ds3 = ds3.add_column('prompt_type', ['financial'] * len(ds3))
ds4 = ds4.add_column('prompt_type', ['therapy'] * len(ds4))
ds = concatenate_datasets([ds1, ds2, ds3, ds4])
# add index to the dataset
ds = ds.add_column('index', list(range(len(ds))))
print(ds[:3])
```

```python
# load the gpt datasets
ds1 = load_dataset('imhann/medical_prompts_responses_gpt', split='train')
ds2 = load_dataset('imhann/legal_prompts_responses_gpt', split='train')
ds3 = load_dataset('imhann/finance_prompts_responses_gpt', split='train')
ds4 = load_dataset('imhann/therapy_prompts_responses_gpt', split='train')
# add column to identify 'prompt_type'
ds1 = ds1.add_column('prompt_type', ['medical'] * len(ds1))
ds2 = ds2.add_column('prompt_type', ['legal'] * len(ds2))
ds3 = ds3.add_column('prompt_type', ['financial'] * len(ds3))
ds4 = ds4.add_column('prompt_type', ['therapy'] * len(ds4))
ds_gpt = concatenate_datasets([ds1, ds2, ds3, ds4])
# add index to the dataset
ds_gpt = ds_gpt.add_column('index', list(range(len(ds_gpt))))
print(ds_gpt[:3])
```

```python
# sample 30 rows from each prompt type
from pprint import pprint
import pandas as pd

df = ds.to_pandas()
sampled_df = df.groupby('prompt_type').apply(lambda x: x.sample(30, random_state=42)).reset_index(drop=True)
pprint(sampled_df.head())
```

```python
# get the same examples from gpt dataset
sampled_ids = sampled_df['index']
df2 = ds_gpt.to_pandas()
sampled_df2 = df2[df2['index'].isin(sampled_ids)]
pprint(sampled_df2.head())
```

```python
# sort sampled_df by index column (to match the order of sampled_df2)
sampled_df = sampled_df.sort_values('index').reset_index(drop=True)
sampled_df.head()
```

```python
# add column for manual evals and save to csv
sampled_df['m_eval'] = 'null'
sampled_df.to_csv('manual_evals_llama.csv', index=False)
sampled_df2['m_eval'] = 'null'
sampled_df2.to_csv('manual_evals_gpt.csv', index=False)
```
